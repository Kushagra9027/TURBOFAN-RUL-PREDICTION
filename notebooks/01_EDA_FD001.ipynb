```python
import pandas as pd

cols = ['engine_id','cycle']

cols += [f'op_setting_{i}' for i in range(1,4)]

cols += [f'sensor_{i}' for i in range(1,22)]

df = pd.read_csv(
    "../data/train_FD001.txt",
    sep=r"\s+",
    header=None,
    names=cols
)

df.head()
```

```python
print(df.shape)
print(df.info())
```

```python
df.isnull().sum()
```

```python
df.engine_id.nunique()
```

```python
import matplotlib.pyplot as plt

life = df.groupby('engine_id')['cycle'].max()

life.hist(bins=20)

plt.xlabel("Engine Life (Cycles)")
plt.ylabel("Count")
plt.title("Distribution of Engine Lifetimes")

plt.show()
```

```python
max_cycle = df.groupby('engine_id')['cycle'].max()

df['RUL'] = df.apply(
    lambda row:
    max_cycle[row.engine_id] - row.cycle,
    axis=1
)


df[['engine_id','cycle','RUL']].head(20)
```

```python
import seaborn as sns

plt.figure(figsize=(15,10))

sns.heatmap(
    df.corr(),
    cmap='coolwarm'
)

plt.show()
```
